```python
import pandas as pd
sal = pd.read_csv("C:\\Users\\Shruti\\Downloads\\Salaries.csv\\Salaries.csv")
#have to convert because dtype is object for basepay so we have to cleanit
sal['BasePay'] = pd.to_numeric(sal['BasePay'], errors='coerce')
sal = sal.dropna(subset=['BasePay'])
sal['BasePay'].mean()
```

```python
#people having chief in their title
def chief_string(title):
    if 'chief' in title.lower():
        return True
    else:
        return False
sum(sal['JobTitle'].apply(lambda x: chief_string(x)))
```

```python
#relation between length of job title and salary
sal['title_len'] = sal['JobTitle'].apply(len)
```
